```python
# ============================================================
# LESSON 6 - TASK 1
# Build and Train a Simple ANN Model
# ============================================================

# 1. Import libraries
import pandas as pd
import numpy as np
import tensorflow as tf

from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from google.colab import files
```

```python
# ============================================================
# 2. Upload CSV file using code
# ============================================================

print("Please select the Mobile_Price_Classification.csv file")

uploaded = files.upload()


# Get the uploaded file name automatically
file_name = list(uploaded.keys())[0]

print("File uploaded successfully:", file_name)
```

```python
# ============================================================
# 3. Read CSV file
# ============================================================

df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# ============================================================
# 4. Display column names
# ============================================================

print("\nColumn names:")
print(df.columns.tolist())

```

```python
# ============================================================
# 5. Separate input features (X) and target (y)
# ============================================================

X = df.drop('price_range', axis=1)
y = df['price_range']


print("\nInput features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# ============================================================
# 6. Split dataset
#    75% Training
#    25% Testing
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
# ============================================================
# 7. Scale the input data
# ============================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


print("\nData scaling completed!")
```

```python
# ============================================================
# 8. Build ANN model
#
#    Input
#       ↓
#    Hidden Layer 1 = 8 neurons
#       ↓
#    Hidden Layer 2 = 4 neurons
#       ↓
#    Output Layer = 1 neuron
# ============================================================

model = keras.Sequential([
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(4, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# ============================================================
# 9. Compile the ANN model
# ============================================================

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# ============================================================
# 10. Display model architecture
# ============================================================

print("\nANN Model Architecture:")
model.summary()
```

```python
# ============================================================
# 11. Train the model
#     Epochs = 100
#     Batch size = 32
# ============================================================

print("\nTraining the ANN model...")

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

```python
# ============================================================
# 12. Evaluate the model
# ============================================================

print("\nEvaluating the model...")

loss, accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
print("Test Accuracy (%):", accuracy * 100)
```

```python
# ============================================================
# 13. Save the model weights
# ============================================================

weights_file = 'mobile_price_ann.weights.h5'

model.save_weights(weights_file)

print("\nWeights saved successfully!")
print("File name:", weights_file)
```

```python
# ============================================================
# 14. Download the weights file automatically
# ============================================================

files.download(weights_file)

print("\nWeights file download started!")
```
